```python
#dependencias
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
```

```python
def cargar_muestrear_csv(path, frac=0.1):
    """
    Función que lee un archivo .csv 
    Guarda el 10% del archivo
    return: DataFrame
    """
    df = pd.read_csv(path, encoding = "latin1")
    df_sample = df.sample(frac=frac, random_state=42)  # 10% de las filas
    return df_sample
```

```python
#lista de rutas a los archivos
files = [
    "../data/Monday-WorkingHours.pcap_ISCX.csv",
    "../data/Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "../data/Friday-WorkingHours-Afternoon-DDoS.pcap_ISCX.csv"
]
```

```python
#cargar y muestrear
datasets = [cargar_muestrear_csv(f) for f in files]

#concatenar en un único DataFrame
df_total = pd.concat(datasets, ignore_index=True)
```

```python
display(datasets[0].columns)
display(datasets[1].columns)
display(datasets[2].columns)
```

### Escribo el dataframe en formato ``Parquet``

```python
#creo una tabla de parquet tomando el DF
tabla = pa.Table.from_pandas(df_total)
```

```python
#creacion del archivo en formato .parquet
pq.write_table(tabla, '../data_sample/subset_dataset.parquet')
print("parquet file successfully")
```

```python
#para confirmar la ejecución, leo la tabla en formato .parquet
tabla_lec = pq.read_table('../data_sample/subset_dataset.parquet')
df_lec = tabla_lec.to_pandas()

print(df_lec)
print(f"\nData types:\n{df_lec.dtypes}")
```
